Encapsulation
***Wrapping up data in a single unit***

````python
class car:
  def __init__(self,year,speed):
    self.year=year
    self._speed=speed
  def getspeed(self):
    print('maximum speed :',self._speed)
  def setspeed(self,speed):
    self._speed=speed

bmw=car(2018,155)
ford=car(2020,140)
bmw.getspeed()
bmw.setspeed(150)
print(bmw._speed)
````

` Polymorphism`
*`Polymorphism is the feature of using the same function in multiple ways`*

````python
class car:
  def __init__(self,name):
    self.name=name
class sedan(car):
  def accelerate(self):
    print(150)
class suv(car):
  def accelerate(self):
    print(180)
obj=[sedan('camry'),suv('scorpio')]
for obj1 in obj:
  print(obj1.name+ ":",end=" ")
  obj1.accelerate()
````

````python
# Draw contour map
import numpy as np
import matplotlib
import matplotlib.cm as cm
import matplotlib.pyplot as plt

delta=0.025
x=np.arange(-3.0,3.0,delta)
y=np.arange(-2.0,2.0,delta)

X,Y=np.meshgrid(x,y)
Z1=np.exp(-X**2-Y**2)
Z2=np.exp(-(X-1)**2-(Y-1)**2)
Z=(Z1-Z2)*2

````

````python
fig,ax=plt.subplots()
CS=ax.contour(X,Y,Z)
ax.clabel(CS,inline=1,fontsize=10)
ax.set_title('contour map')
````
